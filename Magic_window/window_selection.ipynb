# Magic window — 12 export windows

Each window cell **auto-loads setup** — run any window cell directly after a kernel restart.

Edit `window_top`, `window_left`, `window_angle` in that cell's `WINDOW` dict, then re-run.

| # | Export key | Category |
|---|------------|----------|
| 1 | `magicwin_reference` | Reference |
| 2–5 | `magicwin_coast_slide_*` | Coastal slide |
| 6–8 | `magicwin_bathy_*` | Bathymetry |
| 9–12 | `magicwin_rot_*` | Rotation (0°/23°/45°/68°) |

```python
import importlib.util
from pathlib import Path
_p = Path.cwd()
_p = _p if _p.name == "Magic_window" else _p / "Magic_window"
_spec = importlib.util.spec_from_file_location("window_selection_setup", _p / "window_selection_setup.py")
_SETUP = importlib.util.module_from_spec(_spec)
_spec.loader.exec_module(_SETUP)
_SETUP.bootstrap_notebook(globals())

# === Overview — all 12 windows ===
from matplotlib.lines import Line2D

ALL_SPECS = [globals().get(f"SPEC_{i}") for i in range(1, 13)]
ALL_SPECS = [s for s in ALL_SPECS if s is not None]
if len(ALL_SPECS) < 12:
    print(f"Warning: {len(ALL_SPECS)}/12 windows loaded — run missing window cells first.")

fields, t_str = load_fields(VIZ_TIME_IDX)
bathy = fields["bathymetry"]
b_ok = bathy[np.isfinite(bathy) & (bathy != 0)]
b_vmin, b_vmax = float(b_ok.min()), float(b_ok.max())

fig, (ax_map, ax_leg) = plt.subplots(1, 2, figsize=(16, 8), gridspec_kw={"width_ratios": [3, 1.5]})
ax_map.imshow(mask_land(bathy), cmap=bathymetry_cmap(), vmin=b_vmin, vmax=b_vmax, origin="upper")
for i, s in enumerate(ALL_SPECS, start=1):
    _outline(ax_map, s)
ax_map.set_title(f"All windows ({len(ALL_SPECS)}) | t={t_str}")
ax_map.set_xlabel("col"); ax_map.set_ylabel("row")

ax_leg.axis("off")
handles = [
    Line2D([0], [0], color=s["color"], lw=4,
           label=f"{i:2d}. {s['export_key']}\ntop={s['window_top']} left={s['window_left']} {s['window_angle']:.0f}°")
    for i, s in enumerate(ALL_SPECS, start=1)
]
ax_leg.legend(handles=handles, loc="upper left", fontsize=7, title="Export keys")
plt.tight_layout()
plt.show()
```
